```python
import numpy as np
import matplotlib.pyplot as plt

def test_quiver(n_rows, n_cols):

    # grille de points
    x = np.arange(n_cols)
    z = np.arange(n_rows)

    x_2d, z_2d = np.meshgrid(x, z)

    # vecteurs fictifs
    u = np.cos(x_2d / 3)
    v = np.sin(z_2d / 3)

    fig, ax2 = plt.subplots(figsize=(6, 6))
    ax2.set_title(f"Grid {n_rows} x {n_cols}")

    # === CAS : plus de lignes que de colonnes → on transpose ===
    if n_rows > n_cols:
        X_plot = z_2d
        Y_plot = x_2d
        U_plot = v
        V_plot = u
        scale_unit = 'width'
        width = 0.003
    else:
        X_plot = x_2d
        Y_plot = z_2d
        U_plot = u
        V_plot = v
        scale_unit = 'width'
        width = 0.003

    # === Quiver ===
    ax2.quiver(
        X_plot,
        Y_plot,
        U_plot,
        V_plot,
        color='black',
        width=width,
        scale=40,
        scale_units=scale_unit,
    )

    ax2.set_aspect("equal")
    ax2.invert_yaxis()

    plt.tight_layout()
    plt.show()


# ==========================
# Tests
# ==========================
formats = [
    (40,10),
    (10,40),
    (20,20),
    (5,30),
    (30,5),
    (20,30),
    (30,20),
]

for rows, cols in formats:
    test_quiver(rows, cols)
```
