**calcul paramètre nématique**

```python
threshold = 0.01
plot = True
apply_mirror = False
verbose = False



S, mean_angle, dict = compute_nematic_parameter(
    processor = proc,
    qvalue = qvalues,
    threshold = threshold,
    radius = radius,
    L = L,
    radius_pd = radius_pd,
    L_pd= L_pd,
    plot = plot,
    verbose = True
        )
```

**Calcul du paramètre S moyen**

$<S>=w_{//}*S_{//calc} + w_{iso}*S_{iso}$

comme $S_{iso}=0$

$<S>=w_{//}*S_{//calc}$

```python
Savg = S*dict['wparr'] + dict['wiso']*0
print(f"S: {S}, S_avg: {Savg}")
```
